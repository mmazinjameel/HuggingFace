# Login to HuggingFace

```python
from huggingface_hub import login
from dotenv import load_dotenv
import os

load_dotenv()

hf_token = os.getenv("HF_TOKEN")
if hf_token and hf_token.startswith("hf_"):
    print("Good")
    login(token=hf_token, add_to_git_credential=False)

else:
    print("Not Found or invalid token")
```

# Use Pipelines

## Sentiment Analyzer

```python
from transformers import pipeline

my_simple_sentiment_analyzer = pipeline("sentiment-analysis", model = "nlptown/bert-base-multilingual-uncased-sentiment")
result = my_simple_sentiment_analyzer("I am super excited to be on the way to LLM mastery")
print(result)
```

### This model has been downloaded, now we can reuse it many times

```python
result2 = my_simple_sentiment_analyzer("This is terrible and disappointing")
print(result2)
```

# Proper Chat Behavior

```python
from transformers import AutoTokenizer
from transformers import pipeline

tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.1-8B-Instruct")

messages = [
    {"role":"user","content":"Tell me a joke!"}
]

prompt = tokenizer.apply_chat_template(messages, tokenize=False)

pipe = pipeline(task="text-generation", model="meta-llama/Llama-3.1-8B-Instruct", tokenizer=tokenizer)

response = pipe(prompt, max_new_tokens=100)
print(response[0]["generated_text"])
```
